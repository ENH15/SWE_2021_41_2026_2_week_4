### Happy Number
Write an algorithm to determine if a number **n** is happy.

A **happy number** is a number defined by the following process:

* Starting with any positive integer, replace the number by the sum of the squares of its digits.
* Repeat the process until the number equals 1 (where it will stay), or it **loops endlessly in a cycle** which does not include 1.
* Those numbers for which this process **ends in 1** are happy.

Return **true** if n is a happy number, and **false** if not.

Constraints:
1 <= n <= $2^{31}$ - 1

```python
def isHappy(n):
  #take the number n, cut out the last digit while n > 0
  #add it onto sum/temp
  #then when it finishes check if that total is 1
    #if true return 1
    #else call ishappy with sum/temp

  #addition - the top comment has not been changed according to this change
  # forgot ab inf looping. Adding set of seen numbers.
  # come across a seen number - quit return false

  seen = set()

  while (n != 1):
    if n in seen:
      return False

    seen.add(n)

    sum = 0
    while (n > 0):
      last_digit = n % 10
      sum += last_digit ** 2
      n = n // 10
    n = sum

  return True
```
